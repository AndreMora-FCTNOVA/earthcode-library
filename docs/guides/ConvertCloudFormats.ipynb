```python
import os

from rio_cogeo.cogeo import cog_translate
from rio_cogeo.profiles import cog_profiles

import geopandas as gpd
import pyarrow
```

```python

biomassPools = {
"AGBD":"AboveGround Biomass Density Map Geopackage",
"BGBD":"BelowGround Biomass Density Map Geopackage",
"LBD":"Litter Biomass Density Map Geopackage",
"SBD":"Shrub Biomass Density Map Geopackage",

}

metrics = [
    "CV",
    "IQR",
    "KURTOSIS",
    "MEAN",
    "MEDIAN",
    "RELATIVESTANDARDERROR",
    "SKEWNESS",
    "STANDARDERROR",


]
profile = cog_profiles.get("lzw")
```

# Spanish

```python

# Local OSC clone root path (assumed one folder above repository root)
file_root = "C:\Projects\BLS-BIOMASS\Spain-Biomass"

locations = [
    "Extremadura",
    "HuescaZaragoza",
    "Leon",
    "Teruel",
]

```

### **Convert to Cog**

```python
suffix_cog = "FCT-ISA-NMBU_Map.tif"
suffix_cog_out = "FCT-ISA-NMBU_Map.cog"
# convert to COG
for location in locations:
    for biomassPool,biomassPoolDesc in biomassPools.items():
        for metric in metrics:
            filename = location + "_" + biomassPool + "_" + metric +"_"
            cog_translate( os.path.join(file_root,location, filename + suffix_cog),
                           os.path.join(file_root, location, "CloudFormats", filename + suffix_cog_out), profile)
```

### Convert to Parquet

```python
suffix_parquet = "Map.gpkg"
suffix_parquet_out = "Map.parquet"

# convert to parquet
for location in locations:
    for biomassPool,biomassPoolDesc in biomassPools.items():
        filename = location + "_" + biomassPool + "_"

        df = gpd.read_file(os.path.join(file_root, location, filename + suffix_parquet))
        df.to_parquet(os.path.join(file_root, location, "CloudFormats", filename + suffix_parquet_out))

```
